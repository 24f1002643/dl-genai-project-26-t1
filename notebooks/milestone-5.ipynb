End-to-end pipeline validation for AST model

```python
# --- Q1: Dataset Splitting ---
from sklearn.model_selection import train_test_split

genres = 10
recipes_per_genre = 100

recipes = []
for g in range(genres):
    for i in range(recipes_per_genre):
        recipes.append({"genre": g, "id": i})

train_recipes, val_recipes = train_test_split(
    recipes, test_size=0.2, shuffle=True, random_state=42
)

print("Validation set size:", len(val_recipes))
```

```python
# --- Q2: On-the-Fly Mixing Shape ---
import numpy as np

stems = [np.ones(160000) for _ in range(4)]
noise = np.ones(160000)

mix = sum(stems) + 0.2 * noise

print("Mix shape:", mix.shape)
```

```python
# --- Q3: Feature Extractor Shape ---
from transformers import AutoFeatureExtractor

extractor = AutoFeatureExtractor.from_pretrained(
    "MIT/ast-finetuned-audioset-10-10-0.4593"
)

mix = np.ones(160000)

inputs = extractor(mix, sampling_rate=16000, return_tensors="pt")
tensor = inputs["input_values"].squeeze(0)

print("Feature shape:", tensor.shape)
```

```python
# --- Q4: AST Model Parameters ---
from transformers import ASTForAudioClassification

model = ASTForAudioClassification.from_pretrained(
    "MIT/ast-finetuned-audioset-10-10-0.4593",
    num_labels=10,
    ignore_mismatched_sizes=True
)

total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print("Trainable parameters:", total_params)
```

```python
# --- Q5: Normalization ---
y_test = np.array([-0.85, 0.40, 0.20, -0.10])

y_norm = y_test / (np.max(np.abs(y_test)) + 1e-9)

print("Normalized array:", y_norm)
print("Value at index 0:", round(y_norm[0], 3))
```
