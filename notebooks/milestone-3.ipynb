```python
# --- Q1: Flatten Tensor ---
import torch

x = torch.randn(32, 1, 64, 64)
flattened_size = x.view(32, -1).shape[1]
print("Flattened feature size:", flattened_size)
```

```python
# --- Q2: Mel Spectrogram Shape ---
import torchaudio

waveform = torch.randn(1, 16000)

mel_spec = torchaudio.transforms.MelSpectrogram(
    sample_rate=16000,
    n_fft=400,
    hop_length=160,
    n_mels=64
)(waveform)

print("MelSpectrogram shape:", mel_spec.shape)
```

```python
# --- Q3: CNN Output Shape ---
import torch.nn as nn

x = torch.randn(1, 1, 64, 64)

conv = nn.Conv2d(1, 1, kernel_size=3, stride=1, padding=1)
pool = nn.MaxPool2d(kernel_size=2, stride=2)

out = pool(conv(x))
print("Output shape:", out.shape)
```

```python
# --- Q4: Linear Layer Parameters ---
in_features = 128
out_features = 64

weights = in_features * out_features
bias = out_features

total_params = weights + bias
print("Total parameters:", total_params)
```

```python
# --- Q5: Number of Batches ---
import math

dataset_size = 1050
batch_size = 32

batches = math.ceil(dataset_size / batch_size)
print("Total batches:", batches)
```

```python
# --- Q6: CrossEntropy Loss ---
import torch.nn.functional as F

logits = torch.tensor([[2.5, 1.0, 0.1]])
target = torch.tensor([0])

loss = F.cross_entropy(logits, target)
print("Loss:", round(loss.item(), 3))
```

```python
# --- Q7: SGD Update ---
weight = 0.5
grad = 0.2
lr = 0.01

new_weight = weight - lr * grad
print("Updated weight:", new_weight)
```

```python
# --- Q8: Padding Calculation ---
# Formula: padding = (kernel_size - 1) / 2 for same output

kernel_size = 5
padding = (kernel_size - 1) // 2

print("Required padding:", padding)
```
